4. **Обработка выбросов**: [+1 балл]
   - Определить пороги для выбросов на основе статистических мер (например, IQR)
   - Решить, удалять ли выбросы или корректировать их значения.

````python
# Функция для определения выбросов на основе IQR
def detect_outliers_iqr(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return column[(column < lower_bound) | (column > upper_bound)]

# Применяем функцию ко всем числовым колонкам
outliers = {}
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    outliers[col] = detect_outliers_iqr(df[col])

for col, out in outliers.items():
    print(f"Выбросы в колонке '{col}':")
    print(out)
    print()
````

### Что такое IQR и как его использовать для обработки выбросов:

**IQR (Interquartile Range)**, или межквартильный размах, это мера статистического разброса и используется для определения выбросов. IQR рассчитывается как разность между 75-м и 25-м процентилями данных (Q3 и Q1 соответственно).

Выбросы часто определяются как значения, которые находятся за пределами следующих границ:
- Нижняя граница: Q1 - 1.5 * IQR
- Верхняя граница: Q3 + 1.5 * IQR

#### Примерный код на Python с использованием pandas:

```python
import pandas as pd

# Загрузка данных
df = pd.read_csv('path_to_file.csv')

# Рассчитываем IQR для колонки 'example_column'
Q1 = df['example_column'].quantile(0.25)
Q3 = df['example_column'].quantile(0.75)
IQR = Q3 - Q1

# Определяем границы для выбросов
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Фильтрация данных для удаления или корректировки выбросов
filtered_df = df[(df['example_column'] >= lower_bound) & (df['example_column'] <= upper_bound)]
```


````python
import numpy as np
# Обработка выбросов

# Логарифмическое преобразование
df['CRIM'] = np.log1p(df['CRIM'])

# Замена выбросов на медиану
median_zn = df['ZN'].median()
Q1_zn = df['ZN'].quantile(0.25)
Q3_zn = df['ZN'].quantile(0.75)
IQR_zn = Q3_zn - Q1_zn
lower_bound_zn = Q1_zn - 1.5 * IQR_zn
upper_bound_zn = Q3_zn + 1.5 * IQR_zn
df['ZN'] = df['ZN'].apply(lambda x: median_zn if x < lower_bound_zn or x > upper_bound_zn else x)

# Ограничение значений
lower_limit_b = df['B'].quantile(0.05)
upper_limit_b = df['B'].quantile(0.95)
df['B'] = df['B'].clip(lower_limit_b, upper_limit_b)

plt.figure(figsize=(12, 8))

# CRIM
plt.subplot(2, 2, 1)
sns.histplot(df['CRIM'], kde=True)
plt.title('Распределение CRIM после логарифмирования')

# ZN
plt.subplot(2, 2, 2)
sns.histplot(df['ZN'], kde=True)
plt.title('Распределение ZN после замены на медиану')

# B
plt.subplot(2, 2, 3)
sns.histplot(df['B'], kde=True)
plt.title('Распределение B после ограничения значений')

plt.tight_layout()
plt.show()

print("Обработанный DataFrame:")
df
````

6. **Как влияет количество комнат (`RM`) на стоимость жилья (`MEDV`)?** [+0.5 балла]
   - Построить диаграмму рассеяния между `RM` и `MEDV`.
   - Рассчитать коэффициент корреляции Пирсона.
   - Сделать письменный развернутый вывод в отдельной текстовой ячейке: влияет ли количество комнат и на стоимость жилья и почему? Логично ли это?


````python
from scipy.stats import pearsonr

# Построение диаграммы рассеяния между RM и TARGET.
plt.figure(figsize=(10, 6))
plt.scatter(df['RM'], df['TARGET'], alpha=0.5)
plt.title('Диаграмма рассеяния: Количество комнат (RM) vs Стоимость жилья (TARGET)')
plt.xlabel('Количество комнат (RM)')
plt.ylabel('Стоимость жилья (TARGET)')
plt.grid(True)
plt.show()

# Расчет коэффициента корреляции Пирсона
corr, _ = pearsonr(df['RM'], df['TARGET'])
print(f'Коэффициент корреляции Пирсона: {corr:.2f}')
````

Наблюдается высокая положительная корреляция 0,7 между количеством комнат в доме и стоимостью жилья. Конечно, количество комнат это не единственный определяющий фактор стоимости. Большое количество комнат чаще всего указывают на большую площадь дома и его комфорт.

   - В чем разницы корелляции и причинно-следственной связи? Ответить на этот вопрос развернуто в отдельной текстовой ячейке


````python
# Рассчитываем корреляционную матрицу
corr_matrix = df[['LSTAT', 'TARGET']].corr()

# Визуализация тепловой карты
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Тепловая карта корреляции: LSTAT и TARGET')
plt.show()
````

Коэффициент -0,74 означает, что с увеличением доли населения с низким социально-экономическим статусом стоимость жилья уменьшается, и наоборот: чем выше социально-экономический статус населения, тем выше стоимость жилья. Это логично, так как районы с более высоким социально-экономическим статусом обычно характеризуются лучшей инфраструктурой, качеством образования, безопасностью и другими факторами, которые повышают привлекательность жилья и, соответственно, его стоимость.

Коррелция - это мера, которая показывает силу взаимосвязи между переменными, но не указывает на причину данной связи.  Для установления причинно-следственной связи требуется более глубокий анализ, чем просто корреляция.
